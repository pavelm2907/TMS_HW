```python
import numpy as np
```

```python
#операции над матрицами
#1
simple_matrix_a = np.array([
    [1, 2, 4],
    [7, 5, -2],
    [-2, 1, 3]
])

simple_matrix_b = np.array([
    [-1, 0, 3],
    [2, -6, 1],
    [-7, 1, 5]
])

simple_matrix_c = np.array([
    [-2, 3],
    [4, 0],
    [7, -1]
])

def calculate(matr1, matr2, matr3):
    print(f'A + B = \n{matr1 + matr2}')
    print(f'B - A = \n{matr2 - matr1}')
    print(f'A * C = \n{matr1 @ matr3}')
    print(f'A * B * C = \n{matr1 @ matr2 @ matr3}')

calculate(simple_matrix_a, simple_matrix_b, simple_matrix_c)
```

```python
#поиск обратной матрицы
#1
simple_matrix_a = np.array([
    [1, 2, 4],
    [7, 5, -2],
    [-2, 1, 3]
])

def calculate_inv(matr1):
    print(f'Обратная матрица к матрице А:\n{np.linalg.inv(matr1)}')

calculate_inv(simple_matrix_a)
```
