```python
#1
from abc import ABC, abstractmethod
import random as rnd
from typing import List, Dict

class MessageChannel(ABC):

    def __init__(self, message: str, logger: MessageLogger) -> None:
        self._message = message
        self._id = self.generate_id()
        self._logger = logger

    @property
    def text(self) -> str:
        return self._message

    @text.setter
    def text(self, message: str) -> None:
        self._message = message

    @staticmethod
    def generate_id() -> int:
        id_str = ''
        for i in range(5):
            id_str += str(rnd.randint(1, 9))
        return int(id_str)

    @abstractmethod
    def prepare_message(self) -> str:
        pass

    @abstractmethod
    def send_message(self) -> bool:
        pass

class Email(MessageChannel):

    def __init__(self, message, logger):
        super().__init__(message, logger)

    def prepare_message(self) -> str:
        prepared = f'Subject: Notification\n\n{self._message}\n\nBest regards'
        print(prepared)
        return prepared

    def send_message(self) -> bool:
        prepared_message = self.prepare_message()
        print('Email отправлен')
        self._logger.log({
            'ID': self._id,
            'Type': 'Email',
            'Text': prepared_message
        })
        return True

class Sms(MessageChannel):

    def __init__(self, message, logger):
        super().__init__(message, logger)

    def prepare_message(self) -> str:
        prepared = f'SMS: {self._message[:10]}...'
        print(prepared)
        return prepared

    def send_message(self) -> bool:
        prepared_message = self.prepare_message()
        print('SMS отправлено')
        self._logger.log({
            'ID': self._id,
            'Type': 'SMS',
            'Text': prepared_message
        })
        return True

class MessageLogger:
    def __init__(self) -> None:
        self._logs: List[Dict[str, str | int]] = []

    def log(self, message: Dict[str, str | int]) -> None:
        self._logs.append(message)
        print(f"Сообщение с id {message['ID']} залогировано")

    def show_logs(self) -> None:
        print('Логи сообщений\n')
        for log in self._logs:
            print(f"ID: {log['ID']}")
            print(f"Тип: {log['Type']}")
            print(f"Текст: {log['Text']}\n")

logger = MessageLogger()

mes1 = Email('', logger)
mes1.text = 'Желаю тебе всего самого наилучшего'

mes2 = Sms('', logger)
mes2.text = 'Как у тебя дела, что нового ?'

mes1.send_message()
mes2.send_message()

logger.show_logs()
```

```python
#2
class Engine:

    def __init__(self, engine_type: str) -> None:
        self._type = engine_type

    @property
    def type(self) -> str:
        return self._type

    @type.setter
    def type(self, engine_type: str) -> None:
        if engine_type in ['petrol', 'electric']:
            self._type = engine_type
        else:
            print('Incorrect output')

    def start_of_engine(self) -> str:
        if self._type == 'petrol':
            return 'Vroom'
        else:
            return 'Silence'

class Car:

    def __init__(self, model: str, engine_type: str) -> None:
        self._model = model
        self._engine = Engine(engine_type)

    @property
    def model(self) -> str:
        return self._model

    @model.setter
    def model(self, model: str) -> None:
        self._model = model

    @staticmethod
    def is_vin_validate(vin: str) -> bool:
        if len(vin) != 17:
            return False
        elif vin.isalnum():
            return True
        else: return False

    def sound(self) -> str:
       return self._engine.start_of_engine()

car1 = Car('BMW', 'petrol')
car2 = Car('Tesla', 'electric')

print(car1.sound())
print(car2.sound())

Car.is_vin_validate('1HGCM826@3A123456')
```

```python
#3
from abc import ABC, abstractmethod

class PaymentMethod(ABC):

    def __init__(self, amount: float) -> None:
        self._amount = amount

    @property
    def amount(self) -> float:
        return self._amount

    @amount.setter
    def amount(self, amount: float) -> None:
        self._amount = amount

    @abstractmethod
    def calculate_commission(self) -> float:
        pass

    @staticmethod
    def is_sum_valid(amount: float) -> bool:
        if 0 <= amount <= 100:
            return True
        else:
            return False

class BankCard(PaymentMethod):

    def __init__(self, amount):
        super().__init__(amount)

    def calculate_commission(self) -> float:
        return self._amount * 0.1

class EWallet(PaymentMethod):

    def __init__(self, amount):
        super().__init__(amount)

    def calculate_commission(self) -> float:
        return self._amount * 0.05

class TaxCalculator:

    def __init__(self, amount: float) -> None :
        self._amount = amount

    def calculate_tax(self) -> float:
        return self._amount * 0.03

class PaymentProcessor:

    def process_payment(self, payment: PaymentMethod) -> None:
        commission = payment.calculate_commission()
        self._tax = TaxCalculator(payment.amount)
        tax = self._tax.calculate_tax()
        total = commission + tax + payment.amount
        print(f'Сумма: {payment.amount}')
        print(f'Комиссия: {commission}')
        print(f'Налог: {tax}')
        print(f'Итого: {total}\n')


payment1 = BankCard(200)
payment2 = EWallet(200)
processor = PaymentProcessor()
processor.process_payment(payment1)
processor.process_payment(payment2)
PaymentMethod.is_sum_valid(101)
```
