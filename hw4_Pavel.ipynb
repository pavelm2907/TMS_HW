```python
#1
import random
def my_function():
    a = random.randint(1, 100)
    b = random.randint(1, 100)
    sign = random.choice(['+', '-', '*', '/'])
    if sign == '+':
        answer = a + b
    elif sign == '-':
        answer = a - b
    elif sign == '*':
        answer = a * b
    else:
        answer = a / b
    while True:
        print (f'{a} {sign} {b} = ')
        n = int(input())
        if n == answer:
            return 'Правильный ответ'
        else:
            print ('Неправильный ответ')
my_function()
```

```python
#2
import string

my_str: str = 'AbceF'
def find_letter(my_str):
    my_str = my_str.lower()
    my_list = list(my_str)
    alphabet = list(string.ascii_lowercase)
    ind = alphabet.index(my_list[0])
    for char in my_list:
        if char != alphabet[ind]:
            return alphabet[ind]
        else:
            ind += 1
            continue
find_letter(my_str)
```

```python
#3
my_race: list = [60, 80, 50]
def race(my_race: list) -> list:
    if my_race[1] < my_race[0]:
        return None
    time = my_race[2] / (my_race[1] - my_race[0])
    time_sec = time * 3600
    hours = int(time_sec / 3600)
    minutes = int((time_sec % 3600) / 60)
    hours = int(time_sec % 60)
    return [hours, minutes, hours]
race(my_race)
```

```python
#4
def playing_digits(n:int, p:int) -> int:
    n_str = str(n)
    total = 0
    for i, digit_char in enumerate(n_str):
        digit = int(digit_char)
        total += (digit ** (p))
        p += 1
    if total % n == 0:
        return total // n
    else:
        return -1
playing_digits(89, 1)
```

```python
#5
attractions = [
    {"x": 3, "y": 4, "v": 30}
]

hotel_coords = {"x": 0, "y": 0}

start_time = "09:00"
close_time = "10:30"
def is_possible_to_visit_all(attractions, hotel_coords, start_time, close_time):
    time = 0
    for i in range(len(attractions) + 1):
        if i == 0:
            time += ((attractions[i]['x'] - hotel_coords['x'])**2 
            + (attractions[i]['y'] - hotel_coords['y'])**2)**0.5 / 5 * 60
        elif i == (len(attractions)):
            time += ((hotel_coords['x'] - attractions[i - 1]['x'])**2 
            + (hotel_coords['y'] - attractions[i - 1]['y'])**2)**0.5 / 5 * 60
        else:
           time += ((attractions[i]['x'] - attractions[i - 1]['x'])**2 
            + (attractions[i]['y'] - attractions[i - 1]['y'])**2)**0.5 / 5 * 60
    for i in range(len(attractions)):
        time += attractions[i]['v']
    hours, minutes = map(int, start_time.split(':'))
    start = hours * 60 + minutes
    hours, minutes = map(int, close_time.split(':'))
    final = hours * 60 + minutes
    if (final - start) >= time:
        return 'Успели'
    else:
        return 'Не успели'
is_possible_to_visit_all(attractions, hotel_coords, start_time, close_time)
```

```python
#6
start_price_old = 2000
start_price_new = 8000
saving_per_month = 1000
percent_loss_by_month = 1.5
# Ожидается: [6, 766]
def buy_car(start_price_old, start_price_new, saving_per_month, percent_loss_by_month):
    months = 0
    savings = 0
    while True: 
        months += 1
        if months % 2 == 0:
            percent_loss_by_month += 0.5
        start_price_old = start_price_old - (start_price_old * percent_loss_by_month / 100)
        start_price_new = start_price_new - (start_price_new * percent_loss_by_month / 100)
        savings += saving_per_month
        total = start_price_old + savings
        if total >= start_price_new:
            return [months, int(total - start_price_new)]
buy_car(start_price_old, start_price_new, saving_per_month, percent_loss_by_month)     
```

```python
#7
def grabscrab(said, possible_words):
    new_words = []
    for i in range(len(possible_words)):
        count = 0
        if len(said) != len(possible_words[i]):
            continue
        for ch in said:
            count +=  possible_words[i].count(ch)
        if count == len(possible_words[i]):
            new_words.append(possible_words[i])
    return new_words
grabscrab("ab", ["a", "b", "ab", "ba", "abc"])
```

```python
#8
def up_and_down(str_words) -> str:
    str_words = str_words.split()
    while True:
        k = 0
        for i in range(len(str_words) - 1):
            if i % 2 != 0:
                if len(str_words[i]) < len(str_words[i + 1]):                   
                    str_words[i], str_words[i + 1] = str_words[i + 1], str_words[i]
                    k += 1
            elif len(str_words[i]) > len(str_words[i + 1]):
                str_words[i], str_words[i + 1] = str_words[i + 1], str_words[i]
                k += 1
        if k == 0:
            break
    for i in range(len(str_words)):
        if i % 2 != 0:
            str_words[i] = str_words[i].upper()
    str_words = ' '.join(str_words)
    return str_words
up_and_down("after be arrived two My so")
```

```python
#9
def ping_pong(score: str) -> str:
    a, b = map(int, score.split(':'))
    total_score = a + b
    if total_score >= 40:
        if total_score // 2 % 2 == 0:
            return 'first'
        else:
            return 'second'
    if total_score // 5 % 2 == 0:
        return 'first'
    else:
        return 'second'
ping_pong('23:23')
```
