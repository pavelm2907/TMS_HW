```python
class Money:
    
    def __init__(self, rubles = 0, cop = 0, currency = 'bel'):
        self.rubles = rubles
        self.cop = cop
        self.currency = currency

    def balance(self):
        print (f'Валюта: {self.currency}')
        print (f'Сумма на счете {self.rubles},{self.cop} р')
        
    def add_money(self, k, n):
        self.rubles += k
        if n >= 100:
            self.rubles += n // 100
            self.cop += n % 100
        else:
            self.cop += n
        self.balance()
        
    def delete_money(self, k, n):
        total_sum = self.rubles * 100 + self.cop
        total_del = k * 100 + n
        if (total_sum - total_del) < 0:
            return 'Мало денег'
        else:
            self.rubles = (total_sum - total_del) // 100
            self.cop = (total_sum - total_del) % 100
        self.balance()
        
    def divide_money(self, k):
        dividend = int((self.rubles * 100 + self.cop) * k / 100)
        if dividend >= 100:
            self.rubles += dividend // 100
            self.cop += dividend % 100
        else:
            self.cop += dividend
        self.balance()

    def compare_balance(self, other):
        if (self.rubles * 100 + self.cop) > (other.rubles * 100 + other.cop):
            return 'У первого денег больше'
        elif (self.rubles * 100 + self.cop) < (other.rubles * 100 + other.cop):
            return 'У второго денег больше'
        else:
            return 'Баланс одинаковый'
        
tom1 = Money()
tom2 = Money()
tom1.add_money(3, 230)
tom2.add_money(2, 330)
print(tom1.delete_money(5, 40))
tom2.delete_money(1, 20)
tom1.divide_money(30)
tom2.divide_money(4)
tom1.compare_balance(tom2)
```

```python
from typing import Dict

class Bank_Account:
    
    def __init__(self, owner: str, balances: Dict[str, float]) -> None:
        self.owner = owner
        self.balances = balances
        
    def balance(self) -> None:
        print(f'Баланс {self.owner}')
        for key, value in self.balances.items():
            print(f'Валюта: {key} Сумма: {value}')
        print(f'Баланс в BYN: {self.balances["BYN"] + self.balances["USD"] * 2.8 + self.balances["EUR"] * 3.2}')
        print(f'Баланс в USD: {self.balances["BYN"] * 0.4 + self.balances["USD"] + self.balances["EUR"] * 1.2}')
        print(f'Баланс в EUR: {self.balances["BYN"] * 0.3 + self.balances["USD"] * 0.9 + self.balances["EUR"]}')
        print('\n')
        
    def add_byn(self, k: float) -> None:
        self.balances["BYN"] += k
        self.balance()
        
    def add_usd(self, k: float) -> None:
        self.balances["USD"] += k
        self.balance()
        
    def add_eur(self, k: float) -> None:
        self.balances["EUR"] += k
        self.balance()
        
    def delete_byn(self, k: float) -> Optional[str]:
        if (self.balances["BYN"] - k) < 0:
            return 'Не хватает средств'
        else:
            self.balances["BYN"] -= k
        self.balance()

    def delete_usd(self, k: float) -> Optional[str]:
        if (self.balances["USD"] - k) < 0:
            return 'Не хватает средств'
        else:
            self.balances["USD"] -= k
        self.balance()

    def delete_eur(self, k: float) -> Optional[str]:
        if (self.balances["EUR"] - k) < 0:
            return 'Не хватает средств'
        else:
            self.balances["EUR"] -= k
        self.balance()
        
    def transfer_money_byn(self, other: Bank_Account, k: float) -> None:
        self.balances["BYN"] -= k
        other.balances["BYN"] += k
        self.balance()
        other.balance()
        
    def transfer_money_usd(self, other: Bank_Account, k: float) -> None:
        self.balances["USD"] -= k
        other.balances["USD"] += k
        self.balance()
        other.balance()

    def transfer_money_eur(self, other: Bank_Account, k: float) -> None:
        self.balances["EUR"] -= k
        other.balances["EUR"] += k
        self.balance()
        other.balance()

    def transfer_money_byn_to_usd(self, other: Bank_Account, k: float) -> None:
        self.balances["BYN"] -= k
        other.balances["USD"] += k * 0.4
        self.balance()
        other.balance()

    def transfer_money_byn_to_eur(self, other: Bank_Account, k: float) -> None:
        self.balances["BYN"] -= k
        other.balances["EUR"] += k * 0.3
        self.balance()
        other.balance()

    def compare_balance(self, other: Bank_Account) -> str:
        if (self.balances["BYN"] + self.balances["USD"] * 2.8 + self.balances["EUR"] * 3.2) > (other.balances["BYN"] + other.balances["USD"] * 2.8 + other.balances["EUR"] * 3.2):
            return f'У {self.owner} баланс больше'
        elif (self.balances["BYN"] + self.balances["USD"] * 2.8 + self.balances["EUR"] * 3.2) < (other.balances["BYN"] + other.balances["USD"] * 2.8 + other.balances["EUR"] * 3.2):
            return f'У {other.owner} баланс больше'
        else:
            return 'Баланс одинаковый'

ba1 = Bank_Account(owner="Иван Петров", 
                   balances={"BYN": 1000.0, "USD": 150.5, "EUR": 100.0})
ba2 = Bank_Account(owner="Петя Иванов", 
                   balances={"BYN": 2000.0, "USD": 250.5, "EUR": 20.0})
ba1.balance()
ba2.balance()
ba1.add_byn(40.5)
ba2.add_eur(150)
ba1.compare_balance(ba2)
ba1.transfer_money_byn(ba2, 5)
ba2.transfer_money_byn_to_usd(ba1, 500)
```
