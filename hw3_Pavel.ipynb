```python
#1
my_str: str = 'hello world how are you'
def str_to_list(my_str: str) -> list:
    my_list = my_str.split()
    return my_list
str_to_list(my_str)
```

```python
#2
import math as m

time: int = 4
def litres(time: int) -> int:
    lit_of_w = m.floor(time * 0.5)
    return lit_of_w
litres(time)   
```

```python
#3
n: int = 18
def create_list(n: int) -> list:
    new_list = []
    for i in range(n, 0, -1):
        new_list.append(i)
    return new_list
create_list(n)
```

```python
#4
year: int = 2026
def century(year: int) -> int:
    cent = year // 100 + 1
    return cent
century(year)
```

```python
#5
number: int = 123
def digits_numbers(number: int) -> list:
    new_list = []
    for i in range(len(str(number))):
        new_list.append(number % (10))
        number = number // (10)
    return new_list
digits_numbers(number)
```

```python
#6.1
n: int = 3
new_matrix = []
k: int = 0
for i in range(n):
    row = []
    for j in range(n):
        k += 1
        row.append(k)
    if i % 2 != 0:
        row.reverse()
        new_matrix.append(row)
    else:
        new_matrix.append(row)
for row in new_matrix:
    print(row)
```

```python
#6.2
n: int = 3
new_matrix = []
for row in range(n):
    new_matrix.append([])
k = 1
for j in range(n):
    if j % 2 != 0:
        for i in range(n - 1, -1, -1):
            new_matrix[i].append(k)
            k += 1
    else:
        for i in range(n):
            new_matrix[i].append(k)
            k += 1
for row in new_matrix:
    print(row)
```
